## Numpy Array Operations:

```python
import numpy as np
```

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("week 1 temperature:\n",temperatures_w1)
```

```python
print("Shape:",temperatures_w1.shape)
print("Data type:",temperatures_w1.dtype)
print("No.Elements:",temperatures_w1.size)
```

```python
# Step 3: Array operations
```

```python
# A. Convert Celsius to Fahrenheit
temperaturef = (temperatures_w1 * 9/5) + 32
print("Temperatures in Fahrenheit:\n",temperaturef )
```

```python
# B. Find maximum, minimum, and mean
max_tempt = np.max(temperatures_w1)
min_tempt = np.min(temperatures_w1)
mean_tempt = np.mean(temperatures_w1)
print(max_tempt)
print(min_tempt)
print(mean_tempt)
```

```python
# Step 4: Array slicing and indexing
```

```python
# Extract the temperatures for the first three days of the week.
print("Temperatures for the first three days:",temperatures_w1[:3])
# Extract the temperatures for the weekend (last two days)
print("Temperatures for the weekend:",temperatures_w1[5:])
# Extract the temperatures for the middle three days of the week
print("Temperatures for the middle three days:",temperatures_w1[2:5])
```

```python
# Step 5: Create a 2D NumPy array
```

```python
temperatures =np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],[19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
print("Temperatures for Two Weeks:\n",temperatures)
```

```python
# Step 6: Inspect and slice the 2D array
```

```python
print("Shape:",temperatures.shape)
print("Data type:",temperatures.dtype)
print("No.Elements:",temperatures.size)
```

```python
# Extract the temperatures for each week and the weekends (last two days) for both weeks
print("Week 1:", temperatures[0])
print("Week 2:", temperatures[1])
print("Week 1 Weekend:", temperatures[0, -2:])
print("Week 2 Weekend:", temperatures[1, -2:])
```

# Pandas Series:

```python
import pandas as pd
```

```python
# 1. Creating Pandas DataFrame:
```

```python
transactions = pd.DataFrame({"TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
                "ProductCategory":['Electronics','Clothing', 'Electronics', 'Furniture','Clothing', 'Electronics', 'Furniture', 
                'Clothing', 'Furniture', 'Electronics'],
                "Region":['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West','South', 'North'],
                "Amount":[200, 150, 300, 450, 200, 250, 300, 180, 350, 400]})
transactions
```

```python
# 2. Data Exploration:
```

```python
print(transactions.head())
```

```python
print(transactions.tail())
```

```python
print("Shape:",transactions.shape)
print("Column Names:", transactions.columns)
print("Data Types:\n",transactions.dtypes)
```

```python
transactions.info()
```

```python
# Display only the 'ProductCategory' and 'Amount' columns
transactions[['ProductCategory', 'Amount']]
```

```python
# Retrieve the last 3 columns of the DataFrame using iloc or loc
transactions.iloc[:, -3:]
```

```python
# Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200
transactions[(transactions['Region'] == 'North') & (transactions['Amount'] >200 )]
```

```python
 # Find the value counts for the 'ProductCategory' column
transactions['ProductCategory'].value_counts()
```

```python
# Find the unique values in the 'Region' column
transactions['Region'].unique()
```

```python
# Group by 'Region' and find the mean amount for each region
print(transactions.groupby('Region')['Amount'].mean())
```

```python
# 3. Manipulating the DataFrame:
```

```python
# Modify the 'Amount' for the transaction with TransactionID 102 to 165
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
transactions
```

```python
# Add a new column 'Discount' by calculating 10% of the 'Amount'.
transactions['Discount'] = transactions['Amount'] * 0.10
transactions
```

```python
# Remove the row with TransactionID 109
transactions.drop(transactions[transactions['TransactionID'] == 109].index, axis=0, inplace=True)
transactions
```

```python
# Delete the 'Discount' column from the DataFrame
transactions = transactions.drop('Discount',axis=1)
transactions
```
